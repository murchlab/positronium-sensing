# Positronium Sensing: Reproducible Analysis

This notebook is an entrypoint for the analysis workflow in this repository.

## Workflow
Run the notebooks in `notebooks/` in numeric order:

1. `notebooks/01_readout_fidelity_corrected_data.ipynb`
2. `notebooks/02_raw_data.ipynb`
3. `notebooks/03_gate_fidelity_corrected_data.ipynb`
4. `notebooks/04_combined_plots.ipynb`
5. `notebooks/05_readout_integration_weights.ipynb`
6. `notebooks/06_readout_fidelity_correction.ipynb`

Optional:
- `notebooks/A01_integration_weights_training_set.ipynb`

## Inputs and Outputs
- Inputs: `data/` (see `data/README.md`)
- Derived data products: `data_analysis/` (see `data_analysis/README.md`)
- Figure exports: `si_figures_pub/` (see `si_figures_pub/README.md`)
